```python
def calcula_percento(valor, percentual=100):
  quanto = valor * percentual /100
  return quanto

print ('20% de 300 são:{:6.2f}'.format (calcula_percento (300,20)))
print ('100% de 50 são:{:6.2f}'.format (calcula_percento (50)))
```
